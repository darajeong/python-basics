```python
def greet(name):
    print(f"안녕하세요! {name}님")
    print("파이썬 공부를 시작합니다.")

greet("홍길동") #파이썬
greet("홍길래") #SQL
```

```python
# 1. 함수에 파라미터 추가
def greet(name, subject):


# 2. 함수 안에서 파라미터 사용
    print(f"안녕하세요! {name}님")
    print(f"{subject} 공부를 시작합니다.")

# 3. 함수 호출시 파라미터 넘겨주기
greet("홍길동", "파이썬")
greet("홍길래", "SQL")
```

```python
# 두개의 숫자를 입력 받아서 더한 값을 반환하는 함수

# 함수 이름 : first_sum
def first_sum(score1, score2):
    return score1 + score2

num1 = int(input("첫번째 숫자 : "))
num2 = int(input("두번째 숫자 : "))


result = first_sum(num1, num2)

print("결과 :", result)
```

```python
# 두 수(1~100까지)를 입력 받아서 사칙연산 결과를 반환하는 calc라는 함수를 만드시오.
# 4개 결과값을 한번에 반환(Tuple)
# 사용자가 입력한 값이 숫자가 아닐 경우 오류메시지 출력하고 프로그램 종료

def calc(a, b):
    addition = a + b
    subtraction = a - b
    multiplication = a * b
    division = a / b

    return addition, subtraction, multiplication, division


num1 = input("첫 번째 수를 입력하세요: ")
num2 = input("두 번째 수를 입력하세요: ")


if num1.isdigit() and num2.isdigit():
    num1 = int(num1)
    num2 = int(num2)

    if 1 <= num1 <= 100 and 1 <= num2 <= 100:
        result = calc(num1, num2)
        print(result)

    else:
        print("1부터 100까지의 숫자를 입력하세요.")

else:
    print("숫자를 입력해야 합니다.")

```
