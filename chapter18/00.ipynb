```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60


student1 = Student("홍길동", 45)

print(student1)
print(f"{student1.name}님의 점수는 {student1.score}점입니다.")

result = ""

if student1.is_passed():
    result = "당당히 합격"
else:
    result = "아쉽지만 탈락"

print(f"{student1.name}님은 이번 시험에서 {result}하셨습니다.")
```

```python
student1 = Student("홍길동", 75)
student2 = Student("강감찬", 59)
student3 = Student("강감찬", 61)

students = [student1, student2]
students.append(student3)

for student in students:
    print(f"{student.name}님의 점수는 {student.score}점입니다.")

    if student.is_passed():
        result = "당당히 합격"
    else:
        result = "아쉽지만 탈락"

    print(f"{student.name}님은 이번 시험에서 {result}하셨습니다.")
```
