```python
# 1. QUESTION — AI 코드가 오류 없이 실행되면 정답일까요?

def calculate_shipping(total_price):
    if total_price >= 50000:
        return 0
    else:
        return 3000
```

```python
# 구매 합계금액이 50000 이상일 경우 배송비 0
# 구매 합계금액이 50000 미만일 경우 배송비 3000

print(calculate_shipping(49999) == 3000)
print(calculate_shipping(50000) == 0)
print(calculate_shipping(50001) == 0)
```

```python
# 11. LIVE CODING 02 — 수정 후 이전 정상값까지 다시 검사하기
tests = [
    (40000, 3000),
    (49999, 3000),
    (50000, 0),
    (50001, 0),
    (60000, 0),
]
```

```python
for order_amount, expected in tests:
    actual = calculate_shipping(order_amount)
    print(order_amount, actual == expected)
```
